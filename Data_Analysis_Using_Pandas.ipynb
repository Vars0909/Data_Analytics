An exploratory data analysis of the Netflix titles dataset using **pandas**, **matplotlib**, and **seaborn**.

**Dataset:** `netflix_titles.csv` (place it in the same directory as this notebook)

````
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

pd.set_option("display.max_columns", None)
````

# Loading Dataset

````
df = pd.read_csv("netflix_titles.csv")
df.head(2)
````

### Checking Dataset Size

````
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
````

# Data Cleaning

## Checking Missing Values

````
df.isnull().sum()
````

### Handling Numerical Missing Values

````
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
numeric_cols

for col_name in numeric_cols:
    median_value = df[col_name].median()
    df[col_name] = df[col_name].fillna(median_value)
````

### Handling Categorical Missing Values

````
string_cols = df.select_dtypes(include=["object"]).columns.tolist()
string_cols

df[string_cols] = df[string_cols].fillna("Unknown")
````

## Checking for Missing Values Again

````
missing_after = df.isnull().sum()
missing_after[missing_after > 0]
````

# Analysis

## 1. Movies vs TV Shows Count

````
content_type = df["type"].value_counts()
print(content_type)

content_type.plot(kind="bar", color=["#E50914", "#221f1f"])
plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

### 2. Top 10 Countries Producing Content

````
country_analysis = df["country"].value_counts().head(10)
print(country_analysis)

country_analysis.plot(kind="barh", color="#E50914")
plt.title("Top 10 Countries Producing Content")
plt.xlabel("Count")
plt.ylabel("Country")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
````

### 3. Most Common Genres

````
genre_analysis = df["listed_in"].value_counts().head(10)
genre_analysis
````

### 4. Movies Released Per Year

````
year_movie = (
    df[df["type"] == "Movie"]["release_year"]
    .value_counts()
    .sort_values(ascending=False)
    .head(10)
)
year_movie
````

### 5. TV Shows Released Per Year

````
tv_year = (
    df[df["type"] == "TV Show"]["release_year"]
    .value_counts()
    .sort_values(ascending=False)
    .head(10)
)
tv_year
````

## 6. Most Popular Content Rating

````
rating_analysis = df["rating"].value_counts()
print(rating_analysis)

rating_analysis.plot(kind="bar", color="#E50914")
plt.title("Content Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
````

### 7. Content Duration Analysis

````
movie_duration = df[df["type"] == "Movie"]
movie_duration[["duration"]].head(10)
````

### 8. Top Directors on Netflix

````
director_analysis = (
    df[df["director"] != "Unknown"]["director"]
    .value_counts()
    .head(10)
)
director_analysis
````

### 9. Top Actors on Netflix

````
actors = (
    df[df["cast"] != "Unknown"]["cast"]
    .str.split(", ")
    .explode()
)

actor_count = actors.value_counts().head(10)
actor_count
````

### 10. Country-wise Content

````
country_content = (
    df[df["country"] != "Unknown"]["country"]
    .str.split(", ")
    .explode()
)

country_content.value_counts().head(10)
````
